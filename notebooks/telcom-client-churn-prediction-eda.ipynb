```python
#using df.copy() to create a copy of the orginal data and we can work on the cleaned version 

cleaned_df = df.copy()
```

```python
cleaned_df['TotalCharges'] = pd.to_numeric(cleaned_df['TotalCharges'], errors = 'coerce')
```

```python
cleaned_df.info()
```

```python
cleaned_df.isna().sum().sort_values(ascending=False)
```

```python
# Getting the isna rows from the cleaned df 

cleaned_df[cleaned_df.isna().any(axis=1)]
```

```python
#The relationship is that all of them are Tenure they haven't completed their month 
#one way to do it using fillna(0) cleaned_df.fillna(0, inplace=True)

cleaned_df['TotalCharges'] = cleaned_df['tenure']* cleaned_df['MonthlyCharges']
```

```python
cleaned_df.info()
```

```python
cleaned_df.isna().sum().sort_values(ascending=False)
```

```python
cleaned_df.describe()
```

```python
#mapping SeniorCitizen from Yes and No to 0 and and 1

cleaned_df['SeniorCitizen'] = cleaned_df['SeniorCitizen'].map({0 : 'No', 1 : 'Yes'})
```

```python
numerical_cols = cleaned_df.select_dtypes(['float', 'int']).columns.tolist()
```

```python
numerical_cols
```

```python
for ncol in numerical_cols:
    plt.figure()
    sns.histplot(data=cleaned_df, x=ncol, kde=True)
    plt.show
```

```python
for ncol in numerical_cols:
    plt.figure()
    sns.boxplot(data= cleaned_df, x=ncol)
    plt.show
```

```python
for ncol in numerical_cols:
    plt.figure()
    sns.violinplot(data=cleaned_df, x= ncol)
    plt.show()
```

```python
for ncol in numerical_cols:
    plt.figure()
    sns.violinplot(data=cleaned_df, x=ncol)
    sns.pointplot(
        data=cleaned_df,
        x=ncol,
        estimator=np.mean,
        markers='D',
        color='yellow'
    )
    plt.show()
```

```python
for ncol in numerical_cols:
    plt.figure()
    sns.violinplot(data=cleaned_df, x=ncol, y='gender', hue='gender', palette={
        'Male' : 'blue',
        'Female' : 'red'
    })
    sns.pointplot(
        data=cleaned_df,
        x=ncol,
        y='gender',
        estimator=np.mean,
        markers='D',
        color='yellow'
    )
    plt.show()
```

```python
for ccol in categorical_cols:
    plt.figure(figsize=(12,10))
    sns.countplot(data=cleaned_df, x=ccol)
    plt.show()
```

```python
for ccol in categorical_cols:
    plt.figure()
    counts = cleaned_df[ccol].value_counts()
    plt.pie(counts.values , labels=counts.index, autopct='%1.1f%%')
    plt.title(f'Pie Chart of {ccol}')
    plt.show()
```

```python
#Heat map correlation
plt.figure(figsize=(8,5))
sns.heatmap(cleaned_df.corr(numeric_only=True), annot=True, vmin= -1, vmax= 1, cmap= 'coolwarm')
plt.show()
```
